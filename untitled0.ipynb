PROYECTO JARDIN INTELIGENCIA AUTONOMO

```python
# Jardín Inteligente Autónono
# Autor:  Karina Cañar
# Fecha: 02-11-2025
# Simulación de un agente racional basado en sensores y actuadores

import random
import time

# --- Estado Inicial del Entorno ---

def crear_entorno_inicial():
    """Crea el estado inicial del entorno del jardín."""
    return {
        "humedad": random.randint(30, 70),
        "luz": random.randint(100, 700),
        "temperatura": random.randint(10, 35)
    }

# --- Percepción del Entorno---
def percibir_entorno(entorno):
    """Simula la lectura de sensores (devuelve el estado actual del entorno)."""
    return entorno.copy()  # Se devuelve una copia para evitar modificar accidentalmente

# --- Decisión del agente---
def decidir(percepcion):
    """Determina qué acciones deben ejecutarse en base al estado percibido."""
    acciones = []
    h, l, t = percepcion["humedad"], percepcion["luz"], percepcion["temperatura"]

    if h < 40:
        acciones.append("activar_riego")
    if l < 300:
        acciones.append("encender_luz")
    if t > 28:
        acciones.append("activar_ventilador")

    if not acciones:
        acciones.append("sin_accion")

    return acciones

# --- Actuación de agente ---
def actuar(entorno, acciones):
    """Ejecuta las acciones necesarias y devuelve el nuevo estado del entorno."""
    nuevo_entorno = entorno.copy()

    for accion in acciones:
        if accion == "activar_riego":
            nuevo_entorno["humedad"] += 15
            print("💧 Activando riego automático. Humedad actual:", nuevo_entorno["humedad"])
        elif accion == "encender_luz":
            nuevo_entorno["luz"] += 200
            print("☀️ Encendiendo lámparas de crecimiento. Luz actual:", nuevo_entorno["luz"])
        elif accion == "activar_ventilador":
            nuevo_entorno["temperatura"] -= 3
            print("🌬 Activando ventilación. Temperatura actual:", nuevo_entorno["temperatura"])
        elif accion == "sin_accion":
            print("✅ Condiciones óptimas. No se requiere acción.")

    # Ajustar valores dentro de límites realistas
    nuevo_entorno["humedad"] = max(0, min(nuevo_entorno["humedad"], 100))
    nuevo_entorno["luz"] = max(0, min(nuevo_entorno["luz"], 1000))
    nuevo_entorno["temperatura"] = max(0, min(nuevo_entorno["temperatura"], 40))

    return nuevo_entorno

# --- Simulación ---
def simular_agente(ciclos=5):
    """Simula el comportamiento del jardín inteligente durante varios ciclos."""
    entorno = crear_entorno_inicial()

    print("> Bienvenido al Simulador de Jardín Inteligente 🌱")
    print("Estado inicial:", entorno, "\n")

    for ciclo in range(ciclos):
        print(f"--- CICLO {ciclo + 1} ---")
        percepcion = percibir_entorno(entorno)
        acciones = decidir(percepcion)
        entorno = actuar(entorno, acciones)

        # Simular variaciones ambientales externas
        entorno["humedad"] += random.choice([-5, 0, 5])
        entorno["luz"] += random.choice([-100, 0, 100])
        entorno["temperatura"] += random.choice([-1, 1])

        # Reajustar límites después de los cambios
        entorno["humedad"] = max(0, min(entorno["humedad"], 100))
        entorno["luz"] = max(0, min(entorno["luz"], 1000))
        entorno["temperatura"] = max(0, min(entorno["temperatura"], 40))

        print("📊 Estado del entorno tras el ciclo:", entorno, "\n")
        time.sleep(1)

# --- Ejecutar simulación ---
simular_agente(5)
```
